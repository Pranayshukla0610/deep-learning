```python
import numpy as np
import matplotlib.pyplot as plt

from PIL import Image

print("Libraries Imported Successfully")
```

```python
image = np.zeros((100,100,3),dtype=np.uint8)
print(image.shape)
```

```python
print(image)
```

```python
plt.imshow(image)
plt.axis('off')
plt.show()
```

```python
print(image[0,0])
```

```python
print(image[50,50])
```

```python
image[50,50] = [255, 0, 0]
print(image[50,50])
```

```python
plt.imshow(image)
plt.axis('off')
plt.show()
```

```python
image[20, 20] = [255, 0, 0]      # Red
image[20, 30] = [0, 255, 0]      # Green
image[20, 40] = [0, 0, 255]      # Blue

image[20, 50] = [255, 255, 255]  # White
image[20, 60] = [0, 0, 0]        # Black
```

```python
plt.imshow(image)
plt.axis('off')
plt.show()
```

```python
print("Minimum",image.min())
print("Maximum",image.max())
```

```python
gradient = np.zeros((256,256),dtype=np.uint8)
for i in range(256):
  gradient[:,i]= i
```

```python
print(gradient.shape)
```

```python
plt.imshow(gradient, cmap="gray")
plt.axis("off")
plt.show()
```

```python
print("Shape:", gradient.shape)
print("Minimum:", gradient.min())
print("Maximum:", gradient.max())
```

```python
print(gradient[100,0])
```

```python
print(gradient[100, 128])
```

```python
print("Image shape:", image_array.shape)
```

```python
height = image_array.shape[0]
width = image_array.shape[1]
channels = image_array.shape[2]

print("Height:", height)
print("Width:", width)
print("Channels:", channels)
```

```python
height, width, channels = image_array.shape

print("Height:", height)
print("Width:", width)
print("Channels:", channels)
```

```python
print(image_array[0, 0])
```

```python
print(image_array[100, 100])
print(image_array[200, 200])
print(image_array[300, 300])
```

```python
red_channel = image_array[:, :, 0]
green_channel = image_array[:, :, 1]
blue_channel = image_array[:, :, 2]
```

```python
print("Red min:", red_channel.min())
print("Red max:", red_channel.max())

print("Green min:", green_channel.min())
print("Green max:", green_channel.max())

print("Blue min:", blue_channel.min())
print("Blue max:", blue_channel.max())
```

```python
plt.imshow(red_channel, cmap="gray")
plt.title("Red Channel")
plt.axis("off")
plt.show()
```
